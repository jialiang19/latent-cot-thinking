```python
import sys, importlib
from pathlib import Path
sys.path.insert(0, str(Path().resolve().parent))

from pathlib import Path
from src.data.data_processing import Polynomial, Polynomial_CODI, RNG
from src.data.data_processing import TOKEN_DICT as CODI_TOKEN_DICT 
import numpy as np
import torch 
from typing import Dict
```

```python
n_data_per_seq = 5000 
split_ratio = 0.5 

for total_seqs in [3, 4, 8, 10]:
    for mod in [50, 41]: 
        n_data_per_len = [n_data_per_seq for _ in range(total_seqs)]
        split_probs_per_len = [split_ratio for _ in range(total_seqs)] 

        Polynomial_Cot_Gen = Polynomial(mod=mod, save_dir=Path(f'../data/polynomial_data/seq_{total_seqs}/num_{n_data_per_seq}/mod_{mod}/cot'), cot=True)
        Polynomial_NonCot_Gen = Polynomial(mod=mod, save_dir=Path(f'../data/polynomial_data/seq_{total_seqs}/num_{n_data_per_seq}/mod_{mod}/not_cot'), cot=False)
        Polynomial_CODI_Gen = Polynomial_CODI(mod=mod, save_dir=Path(f'../data/polynomial_data/seq_{total_seqs}/num_{n_data_per_seq}/mod_{mod}/codi'), cot=True) 

        Polynomial_Cot_Gen.generate_datafiles(n_data_per_len, split_probs_per_len, rng = np.random.default_rng(0))
        Polynomial_NonCot_Gen.generate_datafiles(n_data_per_len, split_probs_per_len, rng = np.random.default_rng(0))
        Polynomial_CODI_Gen.generate_datafiles(n_data_per_len, split_probs_per_len, rng = np.random.default_rng(0))
```
